# Step 2: Data Inspection and Preparation
This notebook inspects 10,000 synthetic campus climate scenarios. These are generated testing scenarios, not real campus measurements.

```python
import pandas as pd
df = pd.read_csv('../data/processed/cleaned_input.csv')
print('Dataset loaded successfully')
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
```

## Data Preview

```python
display(df.head())
```

## Dataset Structure

```python
print('Data types:')
print(df.dtypes)
print('\nDataset information:')
print(df.info())
```

## Missing Values and Duplicate IDs

```python
print('Missing values:')
print(df.isnull().sum())
print('\nDuplicate scenario IDs:', df['scenario_id'].duplicated().sum())
```

## Numeric Ranges and Statistics

```python
print(df[['temperature_c', 'humidity_pct', 'occupancy_count']].describe())
print('\nTemperature range:', df['temperature_c'].min(), 'to', df['temperature_c'].max())
print('Humidity range:', df['humidity_pct'].min(), 'to', df['humidity_pct'].max())
print('Occupancy range:', df['occupancy_count'].min(), 'to', df['occupancy_count'].max())
```

## Tariff Distribution

```python
print(df['tariff_level'].value_counts().sort_index())
```

## Validation Checks

```python
assert df.shape == (10000, 5)
assert df.isnull().sum().sum() == 0
assert df['scenario_id'].is_unique
assert df['temperature_c'].between(18, 45).all()
assert df['humidity_pct'].between(20, 100).all()
assert df['occupancy_count'].between(0, 100).all()
assert df['tariff_level'].isin(['low', 'medium', 'high']).all()
print('STEP 2 DATA VALIDATION PASSED')
```
